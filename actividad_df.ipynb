```python
import pandas as pd
import sqlite3

conn = sqlite3.connect('Salaries.sqlite')

#Aqui agrega la conexión a la base de datos
df = pd.read_sql_query("SELECT * FROM salaries", conn)

conn.close()
#Crea el dataframe con los datos del datasets
```

1. **Top 10 empleados con mayor salario total (incluyendo beneficios)**
Extrae e imprime un top 10 de expleados con mayor salario incluyendo los beneficios 


```python
# agrupo el top 10 de los empleados con mayor salario imcluyendo beneficios
top_10 = df.groupby('EmployeeName')['TotalPayBenefits'].max().sort_values(ascending=False).head(10).reset_index()

top_10.columns = ['Empleado', 'Salario']

top_10
#todo
```

2. **Filtrar empleados con más de 50,000 en horas extra, la columna a tratar es: (OvertimePay)**

```python
# convierto a numerico los datos de overtimepay
df['OvertimePay'] = pd.to_numeric(df['OvertimePay'], errors='coerce')
# filtro los datos de los empleados con horas extras de trabajo mayores a 50000
empleados_alto_overtime = df[df['OvertimePay'] > 50000].dropna(subset=['OvertimePay'])

# ordeno y muestro los empleados con mayor salario por horas extras de trabajo
resultado = empleados_alto_overtime[['EmployeeName', 'OvertimePay']].sort_values('OvertimePay', ascending=False)
print(resultado)
#todo
```

3. **Contar cuántos empleados únicos hay por año**

```python
# Agrupo y cuento los empleados por año
empleados_por_año = df.groupby('Year')['EmployeeName'].nunique().reset_index()
empleados_por_año.columns = ['año', 'Empleados']

print(empleados_por_año)

#todo
```

4. **Ver cuántos cargos únicos (JobTitle) existen y los 5 más comunes**

```python
# cuento las cargos de los empleados
empleos = df.groupby('JobTitle')['EmployeeName'].count().reset_index()

# ordeno por cantidad de empleados y muestro cuantos cargos hay
empleos.columns = ['Cargos', 'Empleados']
empleos.sort_values('Empleados', ascending=False)
#todo
```

5. **Mostrar el salario total (TotalPay) promedio por año**

```python
# calculo el salario promedio por año
salario_total_año = df.groupby('Year')['TotalPay'].mean().reset_index()

# muesto con el nombre de la columna
salario_total_año.columns = ['año', 'salario promedio']
salario_total_año

#todo
```
